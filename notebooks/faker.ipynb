```python
from faker import Faker
import pandas as pd
```

```python
# С локализацией (русские данные)
fake = Faker("ru_RU")

# С несколькими провайдерами
fake = Faker()
fake.add_provider("faker.providers.person.ru_RU")
fake.add_provider("faker.providers.address.ru_RU")
```

## Локализация

```python
fake_en = Faker("en_US")
fake_ru = Faker("ru_RU")
```

```python
def gen_personal_data(faker: Faker, loc="en") -> pd.DataFrame:
    if loc == "ru":
        person = [
            {
                "full_name": faker.name(),
                "first_name": faker.first_name(),
                "last_name": faker.last_name(),
                "middle_name": faker.middle_name(),
                "gender": faker.random_element(["M", "F"]),
                "age": faker.random_int(18, 90),
                "birthdate": faker.date_of_birth(minimum_age=18, maximum_age=90),
            }
            for i in range(10)
        ]
    else:
        person = [
            {
                "full_name": faker.name(),
                "first_name": faker.first_name(),
                "last_name": faker.last_name(),
                "gender": faker.random_element(["M", "F"]),
                "age": faker.random_int(18, 90),
                "birthdate": faker.date_of_birth(minimum_age=18, maximum_age=90),
            }
            for i in range(10)
        ]
    return pd.DataFrame(person)
```

```python
display(gen_personal_data(fake_ru, loc="ru"))
```

```python
display(gen_personal_data(fake_en))
```

## Банкинг

```python
def gen_bank_data(faker: Faker) -> pd.DataFrame:
    data = [
        {
            "credit_card": faker.credit_card_number(),
            "card_expire": faker.credit_card_expire(),
            "cvv": faker.credit_card_security_code(),
            "card_provider": faker.credit_card_provider(),
            "bank": faker.bank(),
            "bic": faker.bic(),
            "account": faker.bban(),
        }
        for i in range(10)
    ]

    return pd.DataFrame(data)
```

```python
display(gen_bank_data(fake_ru))
```

```python
display(gen_bank_data(fake_en))
```

## Бизнес и компании

```python
def gen_business_data(faker: Faker) -> pd.DataFrame:
    data = [
        {
            "company": faker.company(),
            "company_suffix": faker.company_suffix(),
            "catch_phrase": faker.catch_phrase(),
            "bs": faker.bs(),
            "job": faker.job(),
            "industry": faker.random_element(["IT", "Маркетинг", "Финансы", "Розница"]),
        }
        for i in range(10)
    ]

    return pd.DataFrame(data)
```

```python
display(gen_business_data(fake_ru))
```

## Интернет

```python
def gen_internet_data(faker: Faker) -> pd.DataFrame:
    data = [
        {
            "url": faker.url(),
            "domain": faker.domain_name(),
            "ipv4": faker.ipv4(),
            "ipv6": faker.ipv6(),
            "mac_address": faker.mac_address(),
            "user_agent": faker.user_agent(),
            "username": faker.user_name(),
            "password": faker.password(length=12),
        }
        for i in range(10)
    ]

    return pd.DataFrame(data)
```

```python
display(gen_internet_data(fake_ru))
```

```python
display(gen_internet_data(fake_en))
```
